 |-- _Views: long (nullable = true)
 |-- _WebsiteUrl: string (nullable = true)

```python
# create a spark schema structure tpe using the folloing fields:

from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType

schema = StructType([
    StructField("_AboutMe", StringType(), True),
    StructField("_AccountId", LongType(), True),
    StructField("_CreationDate", TimestampType(), True),
    StructField("_DisplayName", StringType(), True),
    StructField("_DownVotes", LongType(), True),
    StructField("_Id", LongType(), True),
    StructField("_LastAccessDate", TimestampType(), True),
    StructField("_Location", StringType(), True),
    StructField("_Reputation", LongType(), True),
    StructField("_UpVotes", LongType(), True),
    StructField("_Views", LongType(), True),
    StructField("_WebsiteUrl", StringType(), True)
])
```

```python
# We need to parse Posts.xml properly

df = (
    spark.read.format("xml")
    .option("rootTag", "posts")
    .option("rowTag", "row")
    .schema(schema)
    .load("/Volumes/data-plataform-jayzern/default/raw-data/Users.xml")
)
```

```python
display(df.limit(10))
```

```python
# Remove underscore chars of the column titles:
from pyspark.sql.functions import col

new_column_names = {column: column.replace("_", "") for column in df.columns}
df = df.select([col(column).alias(new_column_names[column]) for column in df.columns])
display(df.limit(10))
```
